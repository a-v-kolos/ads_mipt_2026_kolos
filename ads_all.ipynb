```python
# 1. Сделать реверс односвзяного списка.
# Пример:
# Входные данные: 1, 2, 3, 4, 5
# Результат: 5, 4, 3, 2, 1

# сложность по памяти: O(1) потому что мы ничего никуда не копируем
# сложность по времени: O(n) потому что надо пройтись по n элементам


data = [1, 2, 3, 4, 5]

# односвязный список
class ListNode: # нода - пустая ячейка односвязного списка
     def __init__(self, val=0, next=None):
         self.val = val # через точку это типа поле (переменные, хранящиеся внутри класса)
         self.next = next # указатель на следущую ноду

class Solution: # мы не всегда используем конструктор классов, нам он не нужен в этом классе потому что он только для функций
    def reverseList(self, head: ListNode) -> ListNode: # возвращает ListNode
        prev = None # предыдущая нода (None потому что перед головой ничего нет)
        curr = head # нода, которая сейчас

        while curr: # цикл до того момента пока curr не станет None
            nxt = curr.next # запоминаем следующий узел прежде чем разворачивать стрелку
            curr.next = prev # разворот стрелочки обращаемся к значению next внутри curr
            prev = curr # сдвинули prev вперед к текущей ноде
            curr = nxt # сдвигаем curr вперёд, переходим к той ноде, которую до этого сохранили в nxt

        return prev # возвращаем голову

def build_list(values): # строит список из питоновского списка, который массив
    dummy = ListNode() # заглушка перед головой
    tail = dummy # последняя нода, которая пока заглушка
    for v in values:
        tail.next = ListNode(v) # прицепляем новый узел с val=v в конец
        tail = tail.next # сдвигаем хвост на только что добавленную ноду
    return dummy.next # возвращаем настоящую голову после заглушки

def to_pylist(head): # возвращает обратно к виду питоновского списка, иначе мы распечаатаь не сможем
    result = []
    while head: # цикл до того момента пока head не станет None
        result.append(head.val)
        head = head.next
    return result

head = build_list(data)
print(to_pylist(head))                      # [1, 2, 3, 4, 5]

reversed_head = Solution().reverseList(head)
print(to_pylist(reversed_head))             # [5, 4, 3, 2, 1]
```

```python
# сложность по времени: O(d*n + d*log d + K + q2*d), где q2 — число запросов типа 2
# сложность по памяти: O(d*n + d + K), без буфера вывода — O(d*n + d)

# СЛОЖНОСТЬ ПО ВРЕМЕНИ:
#   O(d*n)          — чтение документов и подсчёт score
#   O(d*log d)      — сортировка документов
#   O(K)            — суммарный вывод по всем запросам типа 1 (K = Σk)
#   O(q2 * d)       — запросы типа 2 (pop + insert + обновление pos — O(d) каждый)
#   ИТОГО: O(d*n + d*log d + K + q2*d), где q2 — число запросов типа 2
#
# СЛОЖНОСТЬ ПО ПАМЯТИ:
#   O(d*n)          — хранение всех векторов документов (docs)
#   O(n)            — коэффициенты a
#   O(d)            — items, pos
#   O(K)            — буфер вывода out (можно уменьшить, выводя сразу)
#   ИТОГО: O(d*n + d + K), без буфера вывода — O(d*n + d)

# Пример
# Входные данные
# 2
# 1 100
# 10
# 1 2
# 2 1
# 3 1
# 4 1
# 5 1
# 6 1
# 7 1
# 8 1
# 9 1
# 10 1
# 4
# 1 2
# 1 10
# 2 4 1 1000
# 1 10

# Результат
# 1 10
# 1 10 9 8 7 6 5 4 3 2
# 4 1 10 9 8 7 6 5 3 2


# считываем количество критериев n
n = int(input())

# считываем список коэффициентов a[0..n-1], по которым считается взвешенная сумма документа
a = list(map(int, input().split()))

# считываем количество документов d
d = int(input())

# список исходных векторов документов (значения признаков каждого документа)
docs = []

# список кортежей для сортировки: ((-score, idx), score, idx, doc)
# -score - ключ для сортировки по убыванию score (больше score => меньше -score => раньше в списке)
# idx - индекс документа (нужен для уникальности ключа и стабильного порядка при равных score)
# score - сам вес (взвешенная сумма)
# doc - ссылка на исходный вектор документа
items = []

# pos[i] - текущая позиция документа с индексом i в отсортированном списке items
pos = [0] * d

# считываем все документы и сразу считаем их score
for i in range(d):
    doc = list(map(int, input().split()))  # вектор документа
    docs.append(doc)

    # считаем взвешенную сумму score = Σ a[j] * doc[j]
    s = 0
    for j in range(n):
        s += a[j] * doc[j]

    # добавляем в items кортеж: ключ (-s, i), score s, индекс i, сам документ
    items.append(((-s, i), s, i, doc))

# сортируем по ключу (-score, idx) - по убыванию score, при равенстве - по возрастанию idx
items.sort(key=lambda x: x[0])

# заполнение массива позиций (для каждого документа его индекс в отсортированном items)
for t in range(d):
    pos[items[t][2]] = t

# считываем количество запросов q
q = int(input())

out = []  # накапливаем ответы на запросы типа 1

# выполнение запроса
for _ in range(q):
    z = list(map(int, input().split()))  # читаем запрос

    if z[0] == 1:
        # Запрос типа 1: вывести первые k документов в порядке рейтинга
        k = z[1]
        # выводим индексы первых k элементов items(items[t][2] — 0-based, +1 — 1-based)
        out.append(" ".join(str(items[t][2] + 1) for t in range(k)))

    else:
        # Запрос типа 2: изменить j-й признак i-го документа на значение v
        i_doc = z[1] - 1   # индекс документа (0-based)
        j = z[2] - 1       # индекс признака (0-based)
        v = z[3]           # новое значение

        doc = docs[i_doc]              # ссылка на вектор документа
        delta = a[j] * (v - doc[j])    # на сколько изменится score
        doc[j] = v                     # обновляем признак в исходном документе

        p = pos[i_doc]                 # текущая позиция документа в items
        old_score = items[p][1]        # старый score
        new_score = old_score + delta  # новый score

        # удаление документа из отсортированного списка
        items.pop(p)

        # подготовка нового ключа для вставки: (-new_score, i_doc)
        new_key = (-new_score, i_doc)

        # Бинарный поиск места вставки в отсортированном по ключу списке
        lo, hi = 0, len(items)
        while lo < hi:
            mid = (lo + hi) // 2
            if items[mid][0] < new_key:
                lo = mid + 1
            else:
                hi = mid
        idx = lo

        # вставка документа на найденное место
        items.insert(idx, (new_key, new_score, i_doc, doc))

        # обновление pos для всех элементов, чья позиция сдвинулась
        if p < idx:
            # Документ сдвинулся вправо (вниз по рейтингу):
            # элементы с p до idx сдвинулись на одну позицию влево
            for t in range(p, idx + 1):
                pos[items[t][2]] = t
        else:
            # Документ сдвинулся влево (вверх по рейтингу):
            # элементы с idx до p сдвинулись на одну позицию вправо
            for t in range(idx, p + 1):
                pos[items[t][2]] = t

# Выводим все ответы на запросы типа 1, каждый с новой строки
print("\n".join(out))
```

```python
# q1 — число вставок (append)
# q2 — число поисков (find)
# q3 — число вызовов print_tree
# q4 — число вызовов height
# n — текущее число узлов в дереве


# сложность по времени: O((q1 + q2)·log n + (q3 + q4)·n)
# сложность по памяти: O(q1 · log n) - вставки (append)


# СЛОЖНОСТЬ ПО ВРЕМЕНИ:
#   O(log n)        — append в среднем/худшем для RB-дерева: спуск по дереву (O(log n))
#                     + balance (перекраски O(log n), поворотов не более O(1) на итерацию,
#                     всего O(log n))
#   O(log n)        — find: спуск от корня до листа, каждый шаг отсекает половину
#   O(n)            — print_tree: обход всех n узлов (по одному print на узел)
#   O(n)            — height: рекурсивный обход всех узлов (max по двум поддеревьям)
#   ИТОГО: O(q1·log n + q2·log n + q3·n + q4·n) = O((q1+q2)·log n + (q3+q4)·n)
#
# СЛОЖНОСТЬ ПО ПАМЯТИ:
#   O(n)            — сами узлы дерева (по одному Node на каждую успешную вставку,
#                     плюс один NIL-сентинел)
#   O(1)            — NIL-сентинел (один узел на всё дерево)
#   O(log n)        — стек рекурсии balance (глубина ≤ высоты дерева = O(log n))
#   O(h) = O(log n) — стек рекурсии height / print_tree в худшем случае
#                     (h — высота RB-дерева, h = O(log n))
#   ИТОГО: O(n) — доминируют узлы дерева; вспомогательная память O(log n)
#   (стек рекурсии) не превышает O(n), но при подсчёте «чистой» дополнительной
#   памяти без учёта самих узлов — O(log n).



class Node:
    def __init__(self, value,  left=None, right=None, color='red'):
        self.value = value
        self.color = color
        self.left = left
        self.right = right
        self.parent = None

class Tree:
    def __init__(self):
        self.NIL = Node(value=None, color='black') #создаем узел-заглушку, всегда черный, т.к. листья черные
        self.NIL.left = self.NIL #ссылаем все на NIL, чтобы не делать 100500 проверок is None
        self.NIL.right = self.NIL
        self.NIL.parent = self.NIL

        self.root = self.NIL #пока в дереве ничего нет, корень NIL

    def append(self, value):  #по идее за log(n), потому что каждый раз отсекаем половину
        new_node = Node(value, color='red') #создаем новый узел, по правилу он всегда красный
        new_node.left = self.NIL
        new_node.right = self.NIL
        new_node.parent = self.NIL

        parent = self.NIL           # родитель будущего узла
        current = self.root         # текущий узел, начинаем смотреть с корня
        while current is not self.NIL: #пока не дойдем до пустоты
            parent = current #помним последний реальный узел
            if value < current.value:
                current = current.left
            else:
                current = current.right

        new_node.parent = parent #последний реальный узел становится родителем
        if parent is self.NIL: #возможно только если дерево было пустое
            self.root = new_node
        elif value < parent.value:
            parent.left = new_node
        else:
            parent.right = new_node

        self.balance(new_node)


    def rotation(self, node, direction):
        if direction == 'left':   # Левый поворот: правый ребёнок узла становится на его место, левый внук переездает к деду
            child = node.right                  # будущий корень
            node.right = child.left             # правое поддерево перевешиваем к узлу, вокруг которого поворот
            if child.left is not self.NIL:
                child.left.parent = node
            child.parent = node.parent          # переставляем правого ребенка на место узла, вокруг которого поворот
            if node.parent is self.NIL:
                self.root = child               # возможно это корень всего дерева
            elif node is node.parent.left:     # если нет, то новый корень поддерева подвешиваем туда, где висел узел, вокруг которого мы поворачивали
                node.parent.left = child
            else:
                node.parent.right = child
            child.left = node                   # узел, вокруг которого поворачивали, становится левым ребенком своего парвого ребенка
            node.parent = child
            return child

        else:  # если поворот направо, то левый ребенок узла встанет на его место
            child = node.left
            node.left = child.right
            if child.right is not self.NIL:
                child.right.parent = node
            child.parent = node.parent
            if node.parent is self.NIL:
                self.root = child
            elif node is node.parent.left:
                node.parent.left = child
            else:
                node.parent.right = child
            child.right = node
            node.parent = child
            return child


    def balance(self, node):
        while node.parent.color == 'red': #вставили новый красныц узел и его родитель тоже красный
            parent = node.parent
            grandparent = parent.parent

            if parent is grandparent.left: #новый узел добавили слева от деда
                uncle = grandparent.right

                if uncle.color == 'red': #если дядя красный - перекрашиваем без поворота
                    parent.color = 'black'
                    uncle.color = 'black'
                    grandparent.color = 'red'
                    node = grandparent        # проверяем узлы выше

                else: #дядя черный
                    if node is parent.right: #новый узел справа
                        node = parent
                        self.rotation(node, 'left')   #новый узел становится корнем, а родитель левым ребенком
                        parent = node.parent

                    parent.color = 'black' #перекрашиваем отца
                    grandparent.color = 'red' #перекрашиваем деда
                    self.rotation(grandparent, 'right') #правый поворот вокруг дедушки

            else:        # Наоборот, новый узел добавлен справа от деда
                uncle = grandparent.left

                if uncle.color == 'red': #только перекрашиваем
                    parent.color = 'black'
                    uncle.color = 'black'
                    grandparent.color = 'red'
                    node = grandparent

                else:
                    if node is parent.left: #если узел был добавлен слева, то это зигзаг, который надо распрямить правым поворотом вокруг отца
                        node = parent
                        self.rotation(node, 'right')
                        parent = node.parent

                    parent.color = 'black' #перекрашиваем отца
                    grandparent.color = 'red' #перекрашиваем деда
                    self.rotation(grandparent, 'left') #левы1й поворот вокруг деда

        self.root.color = 'black'


    def print_tree(self, node=None, indent=0): #вывод дерева горизонтально, отступы показывают вложенность, o(n) выводим все узлы
        if node is None:
            node = self.root
        if node is self.NIL:
            return
        self.print_tree(node.right, indent + 1) #правое наверху, каждый слой indent+1, таким образом отделяем слои
        print("      " * indent + str(node.value) + node.color[0]) #пишем узел и цвет
        self.print_tree(node.left, indent + 1) #левое поддерево внизу

    def find(self, value):    #по идее за log(n), потому что каждый раз отсекаем половину
        node = self.root #начинаем поиск с корня
        while node is not self.NIL: #пока не дошли до листьев
            if value == node.value:
                return True
            if value < node.value:
                node = node.left
            else:
                node = node.right
        return False


    def height(self, node=None):
        if node is None:
            node = self.root
        if node is self.NIL:
            return 0
        return max(self.height(node.left), self.height(node.right)) + (node.color == 'black')


t = Tree()
for x in [10, 20, 30, 15, 25, 5, 1, 40, 50, 35]:
    t.append(x)
t.print_tree()

print(t.find(25))
print(t.find(99))

print("height =", t.height())
print("-----------------------------------------------------------------------------")
for x in [11, 17, 21, 2, 36, 70]:
    t.append(x)
t.print_tree()
print("height =", t.height())
```

```python
#время o(n*k) за счет цикла for
#память o(k) два списка по k+1

prices = [3, 2, 6, 5, 0, 3]
k = 2 #количество сделок

#prices =[int(i) for i in input().split()]
#k=int(input())

akc0 = [0] * (k + 1) #тут храним прибыль без акций на руках
akc1 = [-float('inf')] * (k + 1) #тут прибыль с акциями на руках

for price in prices:
    for j in range(k, 0, -1):
        akc0[j] = max(akc0[j], akc1[j] + price)  # либо ничего не делаем, либо продаем акции
        akc1[j] = max(akc1[j], akc0[j - 1] - price) #либо ничего не делаем, либо покупаем акции на деньги, которые были вчера
        print(price, akc0, akc1)
print("Максимальная прибыль:", akc0[k])
```

```python
# n - число гостей (пар заезд-отъезд)
# m - число событий (m = 2n)

# СЛОЖНОСТЬ ПО ВРЕМЕНИ:
#   O(len(s))       — re.findall (один проход по строке-регуляркой)
#   O(n)            — сборка пар guests (цикл по found, len(found) = 2n)
#   O(n)            — построение events (по 2 события на гостя, next_day за O(1):
#                     date.fromisoformat + timedelta)
#   O(n log n)      — events.sort() (сортировка m = 2n элементов)
#   O(n)            — проход по events (m итераций)
#   ИТОГО: O(n log n) доминирует сортировка; память входной строки s не считаем,
#   но если считать — O(len(s)) дополнительно
#
# СЛОЖНОСТЬ ПО ПАМЯТИ:
#   O(n)            — found (2n дат)
#   O(n)            — guests (n пар)
#   O(n)            — events (2n событий)
#   O(1)            — cur, best, best_day
#   ИТОГО: O(n) — доминируют found/guests/events (все линейны по n)

# сложность: все, кроме сортировки, O(n), сортировка O(n log n)
# итого O(n log n) по времени и O(n) по памяти

import re
from datetime import date, timedelta


def next_day(s):
    # "2024-06-20" -> "2024-06-21"
    d = date.fromisoformat(s)
    return str(d + timedelta(days=1))


s = '[("2024-06-01", "2024-06-20"), ("2024-06-05", "2024-06-15"), ("2024-06-08", "2024-06-12"), ("2024-06-10", "2024-06-25"), ("2024-06-10", "2024-06-11"), ("2024-06-10", "2024-06-18"), ("2024-06-12", "2024-06-14"), ("2024-06-15", "2024-06-16"), ("2024-06-22", "2024-06-28")]'
# s = input()

# вытаскиваем все даты подряд, так не важно, какие там кавычки
found = re.findall(r'\d{4}-\d{2}-\d{2}', s)

# собираем обратно пары (заезд, отъезд)
guests = []
for i in range(0, len(found) - 1, 2):
    guests.append((found[i], found[i + 1]))

# каждый гость = два события:
# +1 в день заезда, -1 на следующий день после отъезда
# (в день отъезда гость еще в отеле)
events = []
for check_in, check_out in guests:
    events.append((check_in, 1))
    events.append((next_day(check_out), -1))

# даты в формате ГГГГ-ММ-ДД сортируются как обычные строки
# при одинаковой дате -1 идет раньше +1, т.е. сначала уезжают, потом заезжают
events.sort()

cur = 0       # сколько гостей сейчас
best = 0      # максимум
best_day = None

for day, delta in events:
    cur += delta
    if cur > best:  # строго больше -> берем самый ранний день с максимумом
        best = cur
        best_day = day

if best_day is None:
    print("гостей нет")
else:
    print(best_day, "количество гостей =", best)
```

```python

# V - self.size (число вершин)
# E - число рёбер (len(self.edges))

# время: O(V*E) + O(V*E) + O(V) = O(V*E)
# память: O(V) + O(V) + O(V) + O(E) (хранение self.edges) = O(V + E)


# СЛОЖНОСТЬ ПО ВРЕМЕНИ:
#   O(1)            — add_edge (append в список рёбер)
#   O(V*E)          — bellman_ford (V-1 итерация × E рёбер + O(E) проверка отрицательного цикла;
#                     ранний выход if not upd может дать O(E) в лучшем случае)
#   O(V*E)          — neg_cycle_nums (bellman_ford + поиск ребра-нарушителя O(E)
#                     + прокрутка parent на V шагов + обход цикла O(V))
#   O(V)            — way (проход по parent от target до point, длина ≤ V)
#   ИТОГО: O(V*E) доминирует; при q вызовах — O(q * V*E)
#
# СЛОЖНОСТЬ ПО ПАМЯТИ:
#   O(V + E)        — хранение графа (size + edges)
#   O(V)            — dist, parent в bellman_ford
#   O(V)            — cycle в neg_cycle_nums (длина цикла ≤ V)
#   O(V)            — way (путь длины ≤ V)
#   ИТОГО: O(V + E) — структура графа доминирует над рабочими массивами O(V)




class Graph:
    def __init__(self, size):
        self.size = size
        self.edges = []

    def add_edge(self, u, v, w):
        self.edges.append((u, v, w))

    def bellman_ford(self, point):
        INF = float('inf')

        dist = [INF] * self.size
        parent = [None] * self.size
        dist[point] = 0

        for i in range(self.size - 1):
            upd = False
            for u, v, w in self.edges:
                if dist[u] == INF:
                    continue
                if dist[u] + w < dist[v]:
                    dist[v] = dist[u] + w
                    parent[v] = u
                    upd = True
            if not upd:
                break

        neg = False
        for u, v, w in self.edges:
            if dist[u] == INF:
                continue
            if dist[u] + w < dist[v]:
                neg = True
                break

        return dist, parent, neg

    def neg_cycle_nums(self, point):
        INF = float('inf')
        dist, parent, neg = self.bellman_ford(point)
        if not neg:
            return None

        node = None
        for u, v, w in self.edges:
            if dist[u] == INF:
                continue
            if dist[u] + w < dist[v]:
                node = v
                break

        for i in range(self.size):
            node = parent[node]

        cycle = []
        start = node
        while True:
            cycle.append(node)
            node = parent[node]
            if node == start:
                break
        cycle.append(start)
        cycle.reverse()
        return cycle

    def way(self, parent, point, target):
        if parent[target] is None and target != point:
            return []
        way = []
        node = target
        while node is not None:
            way.append(node)
            if node == point:
                break
            node = parent[node]
        way.reverse()
        return way

'''
g = Graph(4)
g.add_edge(0, 1, 4)
g.add_edge(0, 3, 5)
g.add_edge(1, 2, -3)
g.add_edge(3, 2, 2)

dist, parent, neg = g.bellman_ford(0)
print("Расстояния от 0:", dist)
print("Отрицательный цикл:", neg)
print("Путь 0 -> 2:", g.way(parent, 0, 2))
'''
g2 = Graph(4)
g2.add_edge(0, 1, 1)
g2.add_edge(1, 2, -1)
g2.add_edge(2, 1, -1)
g2.add_edge(0, 3, 5)
g2.add_edge(3, 2, 2)
print("g2.edges:", g2.edges)
dist2, parent2, neg2 = g2.bellman_ford(0)
print("Расстояния:", dist2)
print("Отрицательный цикл:", neg2)
print("Вершины цикла:", g2.neg_cycle_nums(0))
```
